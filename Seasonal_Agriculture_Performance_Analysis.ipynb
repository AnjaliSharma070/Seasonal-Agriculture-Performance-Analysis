#9. Outlier Analysis


#Univariate: Season distribution (Pie Chart)

```python
plt.figure(figsize=(7,7))
df['Season'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, cmap='viridis')
plt.title('Distribution of Records by Season')
plt.ylabel('') #Hide the default 'Season' label on the y-axis
plt.tight_layout()
plt.show()
```

```python
# IQR-based outlier summary for numerical columns

outlier_summary = []

for col in numeric_columns:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower = q1 - 1.5 * iqr
  upper = q3 + 1.5 * iqr
  count = ((df[col] < lower) | (df[col] > upper)).sum()

  outlier_summary.append({
      "Column": col,
      "Lower_Bound": lower,
      "Upper_Bound": upper,
      "Outlier_Count": count,
      "Outlier_Percentage": round(count / len(df) * 100, 2)
  })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
   "Outlier_Count", ascending=False
  )
outlier_summary
```

#13. Additional Student-Driven Analysis

```python
# Student Analysis 1
# Question: Is there a relationship between fertilizer usage and agricultural yield?

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="Fertilizer_kg_ha",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Relationship Between Fertilizer Usage and Agricultural Yield")
plt.xlabel("Fertilizer Used (kg/ha)")
plt.ylabel("Yield (tonnes/ha)")
plt.tight_layout()
plt.show()
```

```python
# Student Analysis 2
# Question: How does disease and pest risk vary across seasons?

season_risk = (
    df.groupby("Season")["Disease_Pest_Risk_pct"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

season_risk.plot(kind="bar")

plt.title("Average Disease and Pest Risk Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Risk (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Student Analysis 3
# Question: Which crop generates the highest average profit margin?

df["Profit_Margin_pct"] = (
    df["Profit_INR"] / df["Revenue_INR"] * 100
)

crop_margin = (
    df.groupby("Crop")["Profit_Margin_pct"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

crop_margin.plot(kind="bar")

plt.title("Average Profit Margin by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Profit Margin (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# 14. Key Insights

### Insight 1 – Kharif shows the strongest overall seasonal performance
**Observation:** Kharif has the highest average yield of 5.63 tonnes/ha and the highest average profit of ₹178,914.65 among the three seasons.  
**Evidence:** General seasonal summary analysis.  
**Interpretation:** Kharif represents the strongest overall agricultural performance in the analyzed dataset.  
**Limitation:** This does not mean that Kharif conditions alone cause higher yield or profit, because several farming and environmental factors may contribute.

### Insight 2 – Zaid has the lowest average yield and profit
**Observation:** Zaid has the lowest average yield at 4.64 tonnes/ha and an average loss of ₹24,804.82.  
**Evidence:** General seasonal summary and seasonal performance analysis.  
**Interpretation:** Agricultural performance during Zaid is comparatively weaker in terms of yield and profitability.  
**Limitation:** The result should not be generalized to all agricultural regions or crops outside this dataset.

### Insight 3 – Zaid uses the highest average amount of water
**Observation:** Zaid has the highest average water usage at 6,419.89 m³, compared with 6,102.20 m³ for Kharif and 5,846.99 m³ for Rabi.  
**Evidence:** Seasonal water-usage analysis and general seasonal summary.  
**Interpretation:** Zaid shows comparatively higher water usage while having the lowest average yield.  
**Limitation:** Higher water usage does not necessarily mean inefficient farming because crop type, climate and irrigation requirements also affect water consumption.

### Insight 4 – Sugarcane has substantially higher yield than other crops
**Observation:** Sugarcane has the highest average yield in every season: 53.46 tonnes/ha in Kharif, 43.29 tonnes/ha in Rabi and 38.42 tonnes/ha in Zaid.  
**Evidence:** Crop-season performance analysis.  
**Interpretation:** Sugarcane consistently records the highest yield among the crops in the dataset across seasons.  
**Limitation:** Yield differences may also reflect crop characteristics and should not be interpreted as an effect of season alone.

### Insight 5 – Irrigation method is associated with differences in yield
**Observation:** Drip irrigation has the highest average yield at approximately 6.58 tonnes/ha, followed by Sprinkler, Flood and Rainfed methods.  
**Evidence:** Irrigation method versus average yield analysis.  
**Interpretation:** The dataset shows an association between irrigation method and agricultural yield, with Drip having the highest average yield.  
**Limitation:** This does not establish that irrigation method alone causes higher yield because other factors may influence the outcome.

### Insight 6 – Disease and pest risk is highest in Kharif
**Observation:** Average disease and pest risk is approximately 54.47% in Kharif, compared with 40.48% in Rabi and 38.22% in Zaid.  
**Evidence:** Student Analysis 2 – Disease and Pest Risk Across Seasons.  
**Interpretation:** Kharif records comparatively higher disease and pest risk in the analyzed data.  
**Limitation:** The analysis shows seasonal association and does not identify the specific environmental or farming factor causing the higher risk.

### Insight 7 – Fertilizer usage does not show a strong relationship with yield
**Observation:** The fertilizer-versus-yield analysis shows no clear strong linear relationship between fertilizer usage and agricultural yield.  
**Evidence:** Student Analysis 1 – Fertilizer Usage vs Agricultural Yield.  
**Interpretation:** Increasing fertilizer usage alone does not appear to consistently correspond with higher yield in this dataset.  
**Limitation:** This does not mean fertilizer has no effect on crops; crop type, soil conditions, irrigation and environmental factors may also influence yield.

### Insight 8 – Profitability varies considerably among crops
**Observation:** Chilli has the highest average profit margin at approximately 33.78%, followed by Sugarcane at approximately 27.35%, while several other crops show negative average profit margins.  
**Evidence:** Student Analysis 3 – Average Profit Margin by Crop and crop-season economic analysis.  
**Interpretation:** Economic performance differs substantially across crops, with Chilli and Sugarcane showing stronger average profitability in the analyzed dataset.  
**Limitation:** These profit-margin results are based on the available dataset and should not be treated as guaranteed profitability for real-world farming decisions.

# 15. Recommendations

### Recommendation 1 – Prioritize seasonal planning based on observed performance
Kharif shows the highest average yield (5.63 tonnes/ha) and average profit (₹178,914.65), while Zaid shows the lowest average yield (4.64 tonnes/ha) and negative average profit (-₹24,804.82). Therefore, seasonal planning should consider the stronger performance observed in Kharif and the weaker performance observed in Zaid. However, these results are based only on the analyzed dataset and should not be treated as a universal rule for all farming conditions.

### Recommendation 2 – Review water usage during Zaid
Zaid has the highest average water usage (6,419.89 m³) but the lowest average yield (4.64 tonnes/ha). Therefore, water usage during Zaid should be reviewed carefully, with attention to improving water-use efficiency. The dataset alone cannot determine the exact reason for the higher water usage.

### Recommendation 3 – Consider crop selection according to seasonal performance
Sugarcane records the highest average yield in all three seasons: 53.46 tonnes/ha in Kharif, 43.29 tonnes/ha in Rabi and 38.42 tonnes/ha in Zaid. Therefore, crop-season planning can consider the observed strong performance of Sugarcane. However, crop selection should also consider local conditions and other factors that are not fully captured by this analysis.

### Recommendation 4 – Give greater attention to disease and pest risk in Kharif
Kharif has the highest average disease and pest risk (54.47%), compared with 40.48% in Rabi and 38.22% in Zaid. Therefore, disease and pest monitoring can be given greater attention during Kharif. The analysis shows an association with season but does not identify the specific cause of the higher risk.

### Recommendation 5 – Evaluate irrigation practices using both yield and efficiency
The analysis shows differences in agricultural performance across irrigation methods. Therefore, irrigation practices should be evaluated using both yield and water-efficiency measures rather than yield alone. The dataset does not establish that one irrigation method is always better under every farming condition.

### Recommendation 6 – Avoid increasing fertilizer usage without evaluating outcomes
The fertilizer-versus-yield analysis does not show a clear strong relationship between fertilizer usage and yield. Therefore, fertilizer decisions should be based on observed crop requirements and outcomes rather than assuming that higher fertilizer usage will automatically produce higher yield. Other factors may also influence agricultural yield.

### Recommendation 7 – Focus economic planning on crops with stronger profitability
The crop-level economic analysis shows considerable variation in profitability, with crops such as Sugarcane and Chilli showing relatively strong average profits in the analyzed dataset. Therefore, economic planning can give greater attention to crops with consistently stronger observed financial performance. These results should not be interpreted as guaranteed future profits because market prices and costs can vary.

### Recommendation 8 – Use data-driven monitoring for future seasonal decisions
The analysis shows meaningful differences in yield, profit, water usage, water efficiency and disease/pest risk across seasons and crops. Therefore, these indicators should be monitored together when making future agricultural plans. The current analysis is based on the available dataset and does not by itself establish causal relationships.

Sugarcane recorded the highest average yield across the analyzed seasons, while disease and pest risk was highest in Kharif at 54.47%. Differences were also observed in water efficiency, irrigation performance, fertilizer usage, and economic outcomes across the dataset.

The most useful analytical findings were the seasonal comparison of yield, production, profit and water usage; crop-season performance comparison; analysis of disease and pest risk; and the relationship between agricultural inputs and outcomes. These findings provide a data-based view of areas that may require further attention in seasonal agricultural planning.

However, the analysis has important limitations. The findings are based only on the available dataset and show patterns or associations rather than proving cause-and-effect relationships. Differences in crop type, environmental conditions, irrigation, farming practices and other factors may influence the observed results. Therefore, the findings should not be generalized to all agricultural regions or farming situations.

Overall, this analysis can support future investigation and decision-making by helping identify seasons, crops, resource-usage patterns and economic outcomes that deserve further study. Future analysis could incorporate more detailed regional data, longer-term observations and additional statistical or predictive techniques to better understand the factors influencing agricultural performance.
